```python

import numpy as np
import matplotlib.pyplot as plt
from hmmlearn import hmm

weather = ["Rainy", "Sunny", "Cloudy"]
observations = ["Umbrella", "No Umbrella", "Raincoat"]

model = hmm.CategoricalHMM(n_components=3)

model.startprob_ = np.array([0.4, 0.3, 0.3])

model.transmat_ = np.array([
    [0.6, 0.2, 0.2],
    [0.2, 0.6, 0.2],
    [0.3, 0.2, 0.5]
])

model.emissionprob_ = np.array([
    [0.7, 0.1, 0.2],
    [0.1, 0.7, 0.2],
    [0.2, 0.2, 0.6]
])

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

matrices = [
    (model.transmat_, "Transition Matrix",
     weather, "Next Weather"),
    (model.emissionprob_, "Emission Matrix",
     observations, "Observation")
]

for ax, (matrix, title, labels, xlabel) in zip(axes, matrices):
    im = ax.imshow(matrix, cmap="Blues", vmin=0, vmax=1)
    ax.set_title(title, fontsize=13)
    ax.set_xticks(range(3), labels, rotation=20)
    ax.set_yticks(range(3), weather)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Hidden Weather")

    for i in range(3):
        for j in range(3):
            ax.text(
                j, i, f"{matrix[i, j]:.0%}",
                ha="center", va="center",
                color="white" if matrix[i, j] > 0.5 else "black"
            )

    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

plt.tight_layout()
plt.show()

X = np.array([[0], [1], [2], [0], [1]])

log_prob, states = model.decode(X, algorithm="viterbi")

print("Observation Sequence:", X.ravel())
print("Predicted Weather:", [weather[s] for s in states])
print("Best Path Log Probability:", round(log_prob, 4))
print("Best Path Probability:", f"{np.exp(log_prob):.8f}")
```
